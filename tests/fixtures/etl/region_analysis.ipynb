# Region analysis

Ad-hoc notebook, mixing PySpark and `%%sql` cells.


```python
from pyspark.sql import functions as F

stores = spark.table('warehouse.dim_store')
regions = spark.table('warehouse.dim_region')
```

```python
%matplotlib inline
!pip install seaborn

store_regions = stores.join(regions, stores.region_id == regions.region_id, 'left')
store_regions.count()
```

```python
%%sql
SELECT c.customer_id, g.region_name
FROM warehouse.dim_customer c
JOIN warehouse.dim_region g
    ON c.region_id = g.region_id
```
